```python
import os

from joblib import load, dump
import pandas as pd
from ydata_profiling import ProfileReport
```

Função para gerar o relatório. Recebe o caminho do arquivo como entrada

```python
def gerador_reporte(caminho) -> None:
    '''
        Gera um relatório de perfilamento para o dataframe especificado
        pelo índice c na lista de dataframes.
        O relatório é salvo como um arquivo HTML.
            Parametros:
                c (int): Índice do dataframe na lista de dataframes.

            Retorna:
                None
    '''
    df = load(caminho)   #carrega arquivo compactado
    display(df)    #mostra arquivo e head
    display(df.head())
    profile = ProfileReport(
                                df,
                                title="Customer Dataset Report",
                                explorative=True
                            )               #cria report da df e salva em html
    profile.to_file(caminho+"_report.html")

    return None
```

O arquivo salvo aqui, em html é um arquivo de marcação! Pegamos ele e colocamos no navegador, que fara a conversao das marcacoes para o
aspecto esperado no relatorio

```python
gerador_reporte('/07_aula_07_df_projeto')
```
